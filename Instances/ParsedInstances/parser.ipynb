## Conversao das instancias para o novo formato

```python
def instance_to_txt(output_path, directed, n, depots, nvehicles, capacity, edges, required_edges): 
    with open(output_path, 'w') as f:
        # direcionado
        f.write(f'{1 if directed else 0}\n')

        # n
        f.write(f'{int(n)}\n')

        # n depositos
        f.write(f'{len(depots)}\n')

        # depositos
        f.write(f'{' '.join(map(str, depots))}\n')

        # veiculos
        f.write(f'{nvehicles}\n')

        # capacidade
        f.write(f'{float(capacity):.6f}\n\n')
        
        # arestas obrigatorias
        f.write(f'{len(required_edges)}\n')
        for u, v in required_edges:
            f.write(f'{u} {v}\n')
        f.write('\n')

        # arestas
        f.write(f'{len(edges)}\n')
        for u, v, w in edges:
            f.write(f'{u} {v} {float(w):.6f}\n')
```

## Leitura das Instancias

### Set A: Dearmon

```python
def read_set_a_instances(output_dir_path, raw_instances_root_path): 
    dearmon_instances_files = os.path.join(raw_instances_root_path, 'dearmon graph files')
    dearmon_instances_info = os.path.join(raw_instances_root_path, 'dearmon graph info')

    processed_instances = 0

    if os.path.exists(dearmon_instances_files) and os.path.exists(dearmon_instances_info):
        
        edges_path          = os.path.join(dearmon_instances_info, 'edges.npy')
        capacity_path       = os.path.join(dearmon_instances_info, 'vehicleCapacity.npy')
        n_path              = os.path.join(dearmon_instances_info, 'numNodes.npy')
        required_edges_path = os.path.join(dearmon_instances_info, 'feasibleRequiredNodes.npy')
        depot_nodes_path    = os.path.join(dearmon_instances_info, 'depotNodes.npy')
        vehicles_path       = os.path.join(dearmon_instances_info, 'totalUAVs.npy')

        edges_list          = list(np.load(edges_path, allow_pickle=True))
        capacity_list       = list(np.load(capacity_path, allow_pickle=True))
        n_list              = list(np.load(n_path, allow_pickle=True))
        req_edges_list      = list(np.load(required_edges_path, allow_pickle=True))
        depot_nodes_list    = list(np.load(depot_nodes_path, allow_pickle=True))
        vehicles_list       = list(np.load(vehicles_path, allow_pickle=True))

        output_dir = os.path.join(output_dir_path, 'SetA')
        os.makedirs(output_dir, exist_ok=True)

        for file in os.listdir(dearmon_instances_files):

            i = int(file.split('.')[0])

            n = n_list[i]

            depot_nodes = depot_nodes_list[i]
            if not hasattr(depot_nodes, '__iter__'):
                depot_nodes = [depot_nodes]

            n_vechicles = vehicles_list[i]
            capacity = capacity_list[i]
            edges = edges_list[i]
            required_edges = req_edges_list[i]

            output_file = os.path.join(output_dir, f'SetA_Instance_{i}.txt')

            instance_to_txt(
                output_path=output_file,
                directed=False,
                n=n,
                depots=depot_nodes,
                nvehicles=n_vechicles,
                capacity=capacity,
                edges=edges,
                required_edges=required_edges
            ) 

            processed_instances += 1
    else:
        print(f'Falha ao processar as instancias do Set A. O caminho para as instancias esta incorreto.')
        return

    print('Set A processado com sucesso!')
    print(f'Instancias processadas: {processed_instances}')
```

```python
read_set_a_instances(output_dir_path=output_dir_path,
                     raw_instances_root_path=raw_instances_root_path)
```

### Set B: Mundo Real sem Vento (Nao Direcionado)

```python
def read_set_b_instances(output_dir_path, raw_instances_root_path):
    rw_info_path = os.path.join(raw_instances_root_path, 'real world graph info')
    rw_files_path = os.path.join(raw_instances_root_path, 'real world graph files/new icy road instance graph')

    if os.path.exists(rw_info_path) and os.path.exists(rw_files_path):
        
        edges_path          = os.path.join(rw_info_path, 'edges.npy')
        capacity_path       = os.path.join(rw_info_path, 'vehicleCapacity.npy')
        n_path              = os.path.join(rw_info_path, 'numNodes.npy')
        required_edges_path = os.path.join(rw_info_path, 'feasibleRequiredNodes.npy')
        depot_nodes_path    = os.path.join(rw_info_path, 'depotNodes.npy')
        vehicles_path       = os.path.join(rw_info_path, 'totalUAVs.npy')

        edges_list          = list(np.load(edges_path, allow_pickle=True))
        capacity_list       = list(np.load(capacity_path, allow_pickle=True))
        n_list              = list(np.load(n_path, allow_pickle=True))
        req_edges_list      = list(np.load(required_edges_path, allow_pickle=True))
        depot_nodes_list    = list(np.load(depot_nodes_path, allow_pickle=True))

        output_dir = os.path.join(output_dir_path, 'SetB')
        os.makedirs(output_dir, exist_ok=True)

        for file in os.listdir(rw_files_path):
            if not file.endswith('.pkl') :
                continue
            
            # Extrai o id da instancia
            i = int(file.split('.')[0].split('US')[-1])

            n = n_list[i]

            depot_nodes = depot_nodes_list[i]
            if not hasattr(depot_nodes, '__iter__'):
                depot_nodes = [depot_nodes]

            capacity = capacity_list[i]
            edges = edges_list[i]
            required_edges = req_edges_list[i]

            output_file = os.path.join(output_dir, f'setb_instance_{i}.txt')

            instance_to_txt(
                output_path=output_file,
                directed=False,
                n=n,
                depots=depot_nodes,
                nvehicles=len(required_edges) // 2, # padrao identificado pelo artigo
                capacity=capacity,
                edges=edges,
                required_edges=required_edges
            )

    else:
        print('Falha ao processar as instancias do Set B. O caminho para as instancias esta incorreto.')
        return

    print('Set B processado com sucesso!')
```

```python
read_set_b_instances(output_dir_path=output_dir_path,
                     raw_instances_root_path=raw_instances_root_path)
```

```python
def read_set_c_instances(output_dir_path, raw_instances_root_path):
    rw_info_path = os.path.join(raw_instances_root_path, 'real world graph info')
    rw_files_path = os.path.join(raw_instances_root_path, 'real world graph files/icy road weather instance graph')
    
    if os.path.exists(rw_info_path) and os.path.exists(rw_files_path):
        
        edges_path          = os.path.join(rw_info_path, 'edges.npy')
        capacity_path       = os.path.join(rw_info_path, 'vehicleCapacity.npy')
        n_path              = os.path.join(rw_info_path, 'numNodes.npy')
        required_edges_path = os.path.join(rw_info_path, 'feasibleRequiredNodes.npy')
        depot_nodes_path    = os.path.join(rw_info_path, 'depotNodes.npy')
        vehicles_path       = os.path.join(rw_info_path, 'totalUAVs.npy')

        edges_list          = list(np.load(edges_path, allow_pickle=True))
        capacity_list       = list(np.load(capacity_path, allow_pickle=True))
        n_list              = list(np.load(n_path, allow_pickle=True))
        req_edges_list      = list(np.load(required_edges_path, allow_pickle=True))
        depot_nodes_list    = list(np.load(depot_nodes_path, allow_pickle=True))
        
        output_dir = os.path.join(output_dir_path, 'SetC')
        os.makedirs(output_dir, exist_ok=True)

        for file in os.listdir(rw_files_path):
            if not file.endswith('.pkl') :
                continue
            
            # Extrai o id da instancia
            i = int(file.split('.')[0].split('US')[-1])

            n = n_list[i]

            depot_nodes = depot_nodes_list[i]
            if not hasattr(depot_nodes, '__iter__'):
                depot_nodes = [depot_nodes]

            capacity = capacity_list[i]
            edges = edges_list[i]
            required_edges = req_edges_list[i]

            output_file = os.path.join(output_dir, f'setb_instance_{i}.txt')

            instance_to_txt(
                output_path=output_file,
                directed=True,
                n=n,
                depots=depot_nodes,
                nvehicles=len(required_edges) // 2, # padrao identificado pelo artigo
                capacity=capacity,
                edges=edges,
                required_edges=required_edges
            )
        
    else:
        print('Falha ao processar as instancias do Set C. O caminho para as instancias esta incorreto.')
        print(os.path.exists(rw_files_path))
        print(os.path.exists(rw_info_path))
        return

    print('Set C processado com sucesso!')
```
